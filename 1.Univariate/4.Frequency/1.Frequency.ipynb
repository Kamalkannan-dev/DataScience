```python
def generateFrequencyDetails(column, dataset):
    freqData = pd.DataFrame(columns=['Unique_values', 'Frequency','Relative_frequency','Cumm_Relative_frequency'])
    
    freqData['Unique_values'] = dataset[column].value_counts().index
    freqData['Frequency'] = dataset[column].value_counts().values
    freqData['Relative_frequency'] = dataset[column].value_counts().values / dataset[column].value_counts().values.sum()
    freqData['Cumm_Relative_frequency'] = freqData['Relative_frequency'].cumsum()
    
    return freqData
```

```python
generateFrequencyDetails('ssc_p', dataset)
```

```python
#For Continous numerical data Frequency analysis is not useful, because unique values will be very less
#due to decimal and small differences
#Analysing data in Ranges will be very useful.

#Define the ranges, the data needs to be segregated
ssc_p_bins = [30,40,50,60,70,80,90,100] #30-40, 40-50, ....90-100

#categorize data into the defined ranges
categories_ssc_p = pd.cut(dataset['ssc_p'], ssc_p_bins)
categories_ssc_p
```

```python

# Add the categorized column of ssc_p in dataset
dataset['categories_ssc_p'] = categories_ssc_p

dataset
```

```python
#Perform frequency analysis for the categorized column
generateFrequencyDetails('categories_ssc_p', dataset)
```
