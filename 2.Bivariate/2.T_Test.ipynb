```python
import pandas as pd
import numpy as np
import warnings
from Bivariate import Bivariate
warnings.filterwarnings("ignore")

dataset = pd.read_csv("cleaned_ckd_dataset.csv")

dataset.drop('id',inplace=True, axis=1)

quan, qual = Bivariate.quanQual(dataset)
```

### One Sample T-test
#### Test -> Average of "wc" is 8000
#### H0 - Average = 8000
#### H1 - Average different from 8000

```python
from scipy import stats

hypo_average = 8000

t_stat, p_val = stats.ttest_1samp(dataset['wc'], hypo_average)

print(f'Mean of wc : {dataset['wc'].mean(): .2f}')
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.3f}")
if p_val < 0.05:
    print("Reject H0: Average wc is significantly different from 8,000.")
else:
    print("Fail to reject H0: No significant difference from 8,000.")
```

### Independent 2 sample T-test (unpaired)
#### Test -> Check similarity of (ckd and notckd) vs pot
#### H0 - Avg pot levels of ckd = Avg pot levels of notckd
#### H1 - Avg pot levels of ckd != Avg pot levels of notckd

```python
from scipy import stats

ckd_pot = dataset[dataset['classification']=='ckd']['pot']
notckd_pot = dataset[dataset['classification']=='notckd']['pot']

t_stat, p_val = stats.ttest_ind(ckd_pot, notckd_pot)

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.3f}")
if p_val < 0.05:
    print("Reject H0: Average pot of ckd is significantly different from notckd.")
else:
    print("Fail to reject H0: No significant difference in pot of ckd and notckd.")
```

### Paired sample T-test (Dependent)
#### Test -> Check similarity of ckd for "sod" and "pot"
#### H0 -> Avg of "sod" of ckd = Avg of "pot" of ckd
#### H1 -> Avg of "sod" of ckd != Avg of "pot" of ckd

```python
from scipy import stats

ckd_sod = dataset[dataset['classification']=='ckd']['sod']
ckd_pot = dataset[dataset['classification']=='ckd']['pot']

t_stat, p_val = stats.ttest_rel(ckd_sod, ckd_pot)

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.3f}")
if p_val < 0.05:
    print("Reject H0: Average sod of ckd is significantly different from Average pot of ckd.")
else:
    print("Fail to reject H0: No significant difference in pot and sod of ckd.")
```

#### Comparing means of sod and pot of ckd is meaningless. they are different units biologically, so the mean will always be different, A good example  for paired sample t-test is before vs after of same entity eg: Student score before vs after coaching.
